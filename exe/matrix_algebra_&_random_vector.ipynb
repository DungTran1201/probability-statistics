```python
# Problem 38 
df = pd.read_csv(root / "data" / "Table2_AoLen_Pullover.csv")

print("Dữ liệu mẫu" + 50 * "-" )
print(df.head(5))

print("trung bình X_i" + 40 * "-") 
print(df.drop(columns =["STT"]).mean())

print ("Hiệp phương sai mẫu" +  40 * "-")
print(df.drop(columns =["STT"]).cov())
    

print("Ma trận Phân hoạch" + 40 * "-")
S_n = df.drop(columns =["STT"]).cov().to_numpy()
# Khối S11: Hàng 0,1 & Cột 0,1 (ứng với X1, X2)
S11 = S_n[0:2, 0:2]

# Khối S22: Hàng 2,3 & Cột 2,3 (ứng với X3, X4)
S22 = S_n[2:4, 2:4]

# Khối S12: Hàng 0,1 & Cột 2,3 (hiệp phương sai giữa {X1,X2} và {X3,X4})
S12 = S_n[0:2, 2:4]

# Khối S21: Hàng 2,3 & Cột 0,1 (chuyển vị của S12)
S21 = S_n[2:4, 0:2]

# In kết quả các khối
print("Khối S11 (Hiệp phương sai giữa X1 và X2):")
print(np.round(S11, 4))

print("\nKhối S22 (Hiệp phương sai giữa X3 và X4):")
print(np.round(S22, 4))

print("\nKhối S12 (Hiệp phương sai giữa {X1, X2} và {X3, X4}):")
print(np.round(S12, 4))


print("Ma trận hệ số tương quan" + 40 * "-") 
print(df.drop(columns =["STT"]).corr())
```
