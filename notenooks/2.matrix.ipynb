### ***1.1. Ma trận trực giao (Orthogonal Matrix)***

- **Định nghĩa:** Một ma trận vuông $Q$ được gọi là ma trận trực giao nếu thỏa mãn điều kiện:
  $$QQ' = Q'Q = I \quad \text{hay} \quad Q' = Q^{-1}$$
- **Tính chất:** Các dòng (hoặc các cột) của ma trận trực giao là các véc tơ có độ dài bằng $1$ (đã chuẩn hóa) và đôi một vuông góc với nhau. Tức là, gọi $q_i$ là dòng thứ $i$, ta có $q_i'q_i = 1$ và $q_i'q_j = 0$ (với $i \neq j$).

#### Exercise 14. Kiểm tra ma trận trực giao

```python
# Khai báo ma trận Q
Q = np.array([
    [5/13, 12/13],
    [-12/13,  5/13]
])

# Tính Q_T * Q
Q_T = Q.T
product = np.dot(Q_T, Q)

# In kết quả
print("Ma trận Q:")
print(Q)
print("\nMa trận chuyển vị Q^T:")
print(Q_T)
print("\nKết quả Q^T * Q:")
print(np.round(product, 8)) # Làm tròn để tránh sai số số thực

# Kiểm tra điều kiện ma trận đơn vị
is_orthogonal = np.allclose(product, np.eye(2))
print(f"\nQ có phải ma trận trực giao không? -> {is_orthogonal}")
```

###  **3. Giá trị riêng và Véc tơ riêng (Eigenvalues and Eigenvectors)**

- **Định nghĩa:** Cho ma trận vuông $A$. Một số $\lambda$ được gọi là giá trị riêng của $A$, ứng với véc tơ riêng $x \neq 0$, nếu thỏa mãn phương trình:
  $$Ax = \lambda x$$
- **Tính chất ma trận đối xứng:** Nếu $A$ là ma trận vuông đối xứng cấp $k \times k$, nó sẽ có $k$ cặp giá trị riêng và véc tơ riêng $(\lambda_i, e_i)$. Các véc tơ riêng có thể được chuẩn hóa ($e_i'e_i = 1$) và luôn trực giao với nhau ($e_i'e_j = 0$ khi $i \neq j$).

- **Phương trình đặc trưng:** 
Giá trị riêng $\lambda$ thỏa mãn phương trình đặc trưng 
$$det(A - \lambda I) = 0 $$

Sau đó giải phương trình tìm giá trị của $\lambda$ và thay $\lambda$ vào phương trình đặc trưng nhân với vector $X$ và giải ra các giá trị $x_1, x_2, ...., x_n $ là tìm được vector riêng của ma trận

### **4. Dạng toàn phương và Ma trận xác định dương**

- **Dạng toàn phương (Quadratic Form):** Biểu thức có dạng $x'Ax$, trong đó $A$ là ma trận đối xứng.
- **Xác định không âm (Non-negative definite):** Ma trận đối xứng $A$ (hoặc dạng toàn phương tương ứng) là xác định không âm nếu $x'Ax \geq 0$ với mọi véc tơ $x$. Điều này xảy ra khi và chỉ khi mọi giá trị riêng của $A$ đều không âm ($\lambda_i \geq 0$).
- **Xác định dương (Positive definite):** Ma trận đối xứng $A$ là xác định dương nếu $x'Ax > 0$ với mọi véc tơ $x \neq 0$. Điều này tương đương với việc mọi giá trị riêng của $A$ đều dương ($\lambda_i > 0$).
- **Ý nghĩa hình học:** Dạng toàn phương xác định dương $x'Ax$ đại diện cho bình phương của khoảng cách thống kê từ điểm $x$ tới gốc tọa độ. Tập hợp các điểm có khoảng cách hằng số $c$ tạo thành các hình elip (hoặc siêu elip) có độ dài các bán trục tỷ lệ với $1/\sqrt{\lambda_i}$.

```python
from core.matrix.base import Matrix
from core.matrix.analysis import MatrixAnalysis
A = Matrix([[9, -2],[-2, 6]])
analysis = MatrixAnalysis(A)
print("=" * 60)
print("MATRIX A")
print("=" * 60)
A_T = A.transpose()
is_symmetric = np.allclose(A.data,A_T.data)
print("\nA symmetric:", is_symmetric)

eigenvalues_A = np.linalg.eigvalsh(A.data)
print("Eigenvalues của A:")
for i, value in enumerate(eigenvalues_A, start=1):
    print(f"λ{i} = {value:.6f}")
is_positive_definite = np.all(eigenvalues_A > 0)
print("\nTất cả eigenvalues > 0:",is_positive_definite)
if is_symmetric and is_positive_definite:
    print("\n=> A là positive definite.")
else:
    print("\n=> A không phải positive definite.")

eigenvalues_A, eigenvectors_A = np.linalg.eigh(A.data)
print("Eigenvalues:")
for i, value in enumerate(eigenvalues_A, start=1):
    print(f"λ{i} = {value:.6f}")
print("\nEigenvectors:")
for i in range(len(eigenvalues_A)):
    eigenvalue = eigenvalues_A[i]
    eigenvector = eigenvectors_A[:, i]
    print(f"\nEigenvalue λ{i + 1} = {eigenvalue:.6f}")
    print("Eigenvector:")
    print(eigenvector)

Q = Matrix(eigenvectors_A)
Lambda = Matrix(np.diag(eigenvalues_A))
print(f"Q = {Q}")
print(f"\nΛ = {Lambda}")
print(f"\nQ^T ={Q.transpose()}")
A_reconstructed = (Q@ Lambda@ Q.transpose())
print(f"\nQΛQ^T = {A_reconstructed}")
print("\nKiểm tra A = QΛQ^T:", np.allclose(A.data,A_reconstructed.data))

A_inverse = A.inverse()
print(f"A^(-1) = {A_inverse}")
identity_check = (A@ A_inverse)

print(f"\nA × A^(-1) = {identity_check}")
print("\nKiểm tra bằng Identity:",np.allclose(identity_check.data,np.eye(A.rows)))


eigenvalues_inverse, eigenvectors_inverse = (np.linalg.eigh(A_inverse.data))
print("Eigenvalues của A^(-1):")
for i, value in enumerate(eigenvalues_inverse,start=1):
    print(f"λ{i} = {value:.6f}")
print("\nEigenvectors của A^(-1):")
for i in range(len(eigenvalues_inverse)):
    eigenvalue = eigenvalues_inverse[i]
    eigenvector = eigenvectors_inverse[:, i]
    print(f"\nEigenvalue λ{i + 1} = " 
          f"{eigenvalue:.6f}")
    print(f"Eigenvector:{eigenvector}")

print(f"1 / eigenvalues(A): {1 / eigenvalues_A}")
print(f"\neigenvalues(A^(-1)): {eigenvalues_inverse}")

print("\nKiểm tra:", np.allclose(1 / eigenvalues_A,eigenvalues_inverse))
```
