### **1. Hệ vector độc lập và phụ thuộc tuyến tính**

- **Phụ thuộc tuyến tính:** Một hệ các véc tơ $x_1, x_2, \dots, x_k$ cùng chiều được gọi là phụ thuộc tuyến tính nếu tồn tại các hằng số $c_1, c_2, \dots, c_k$ không đồng thời bằng không sao cho:
  $$c_1x_1 + c_2x_2 + \dots + c_kx_k = 0$$
  Điều này có nghĩa là có ít nhất một véc tơ trong hệ có thể biểu diễn dưới dạng tổ hợp tuyến tính của các véc tơ còn lại.
- **Độc lập tuyến tính:** Hệ các véc tơ được gọi là độc lập tuyến tính nếu chúng không phụ thuộc tuyến tính (đẳng thức trên chỉ xảy ra khi tất cả $c_i = 0$).

### **2. Hình chiếu của một vector**

Cho hai véc tơ $x$ và $y$. 
- **Công thức hình chiếu:** Hình chiếu của véc tơ $x$ lên véc tơ $y$ được xác định bởi:
  $$\text{Hình chiếu} = \left(\frac{x'y}{y'y}\right)y $$
- **Độ dài hình chiếu:** 
  $$\text{Độ dài} = \frac{|x'y|}{L_y} = L_x|\frac{x'y}{L_xL_y}|=  L_x |\cos(\theta)|$$
  Trong đó, $L_x$ và $L_y$ lần lượt là độ dài của véc tơ $x$ và $y$, và $\theta$ là góc tạo bởi hai véc tơ.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Khai báo 2 véc-tơ x và y
x = np.array([3, 4])
y = np.array([4, 0])

# 2. Tính toán theo công thức
x_dot_y = np.dot(x, y)
y_dot_y = np.dot(y, y)

# Véc-tơ hình chiếu
proj_y_x = (x_dot_y / y_dot_y) * y

# Độ dài các véc-tơ
L_x = np.linalg.norm(x)
L_y = np.linalg.norm(y)
len_proj = abs(x_dot_y) / L_y

# In kết quả tính toán
print(f"Véc-tơ x: {x}")
print(f"Véc-tơ y: {y}")
print(f"Véc-tơ hình chiếu của x lên y: {proj_y_x}")
print(f"Độ dài véc-tơ x (L_x): {L_x}")
print(f"Độ dài véc-tơ y (L_y): {L_y}")
print(f"Độ dài hình chiếu: {len_proj}")

# 3. Vẽ đồ thị trực quan
plt.figure(figsize=(8, 6))

# Vẽ trục tọa độ
plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)

# Vẽ véc-tơ x, y và hình chiếu
plt.quiver(0, 0, x[0], x[1], angles='xy', scale_units='xy', scale=1, color='blue', label=r'Véc-tơ $\mathbf{x} = [3, 4]^\top$')
plt.quiver(0, 0, y[0], y[1], angles='xy', scale_units='xy', scale=1, color='green', label=r'Véc-tơ $\mathbf{y} = [4, 0]^\top$')
plt.quiver(0, 0, proj_y_x[0], proj_y_x[1], angles='xy', scale_units='xy', scale=1, color='red', label=r'Hình chiếu $\mathrm{Proj}_{\mathbf{y}}(\mathbf{x}) = [3, 0]^\top$')

# Vẽ đường nét đứt chiếu vuông góc từ mút véc-tơ x xuống hình chiếu
plt.plot([x[0], proj_y_x[0]], [x[1], proj_y_x[1]], 'k--', label='Đường chiếu vuông góc')

# Thiết lập tỷ lệ khung hình 1:1
plt.xlim(-1, 6)
plt.ylim(-1, 5)
plt.gca().set_aspect('equal', adjustable='box')
plt.grid(True, linestyle=':', alpha=0.6)
plt.xlabel('Trục $X_1$')
plt.ylabel('Trục $X_2$')
plt.title('Trực quan hóa Hình chiếu của Véc-tơ $\mathbf{x}$ lên Véc-tơ $\mathbf{y}$')
plt.legend(loc='upper right')
plt.show()
```

###  **3. Giá trị riêng và Véc tơ riêng (Eigenvalues and Eigenvectors)**

- **Định nghĩa:** Cho ma trận vuông $A$. Một số $\lambda$ được gọi là giá trị riêng của $A$, ứng với véc tơ riêng $x \neq 0$, nếu thỏa mãn phương trình:
  $$Ax = \lambda x$$
- **Tính chất ma trận đối xứng:** Nếu $A$ là ma trận vuông đối xứng cấp $k \times k$, nó sẽ có $k$ cặp giá trị riêng và véc tơ riêng $(\lambda_i, e_i)$. Các véc tơ riêng có thể được chuẩn hóa ($e_i'e_i = 1$) và luôn trực giao với nhau ($e_i'e_j = 0$ khi $i \neq j$).
